```python
import numpy as np
import matplotlib.pyplot as plt

import pandas as pd
```

```python
exp_ids = [16,15,17,18]#, 4,5,6,7,9]
plt.figure(dpi=300)


clr = ['r','g','y','b']
id = 0
for exp_id in exp_ids:
    df = pd.read_csv(f'PID_log_{exp_id}.csv')

    # df['time'] = pd.to_datetime(df['#time'])

    dt_fix = df['#time'].to_numpy()

    


    set_point = df['set'].to_numpy()==933

    idx = np.flatnonzero(set_point)[0]

    end_point = df['set'].to_numpy()==4199
    indx_end = np.flatnonzero(end_point[idx:])[0]
    
    print(idx)

    plt.subplot(211)

    plt.plot(dt_fix[idx:(indx_end+idx)]-dt_fix[idx],df['enc'][idx:(indx_end+idx)],'.-',color=clr[id])
    # plt.plot(dt_fix,df['set'],'--',color=clr[id])
    # plt.plot(df['time']-df['time'][0],df['set'],'.',color=clr[id])
    plt.ylabel('enc')
    plt.xlabel(r'$\Delta t$')
    # plt.xlim([0,400])

    plt.subplot(212)
    plt.plot(dt_fix[idx:(indx_end+idx)]-dt_fix[idx],df['speed'][idx:(indx_end+idx)],'.--',color=clr[id],label=f'exp:{exp_id}')
    plt.ylabel('motor speed')
    plt.xlabel(r'$\Delta t$')
    plt.tight_layout()
    # plt.xlim([0,400])

    id=id + 1

plt.legend()
```
